# Plotting Reaction Profiles

This notebook introduces data visualization techniques for chemical reaction engineering.

The focus is on using Matplotlib to visualize:

- concentration profiles
- conversion profiles
- reaction rate profiles
- kinetic relationships
- Arrhenius behavior

## 0. Setup

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Why visualization matters in reaction engineering

Reaction engineering models often generate large amounts of numerical data.

Visualization helps engineers understand:

- how concentration changes with time
- how conversion develops inside reactors
- how temperature affects reaction rates
- how kinetic models compare with experiments

## 2. Basic plotting with reaction data

```python
time = np.array([0, 5, 10, 15, 20])

CA = np.array([
    1.00,
    0.82,
    0.67,
    0.55,
    0.45
])
```

```python
plt.figure(figsize=(8,4))

plt.plot(
    time,
    CA,
    marker="o",
    label="CA"
)

plt.xlabel("Time")
plt.ylabel("Concentration of A")
plt.title("Concentration profile")

plt.legend()
plt.grid(True)

plt.show()
```

## 3. Concentration-time profiles

For a first-order batch reaction:

$$
-\frac{dC_A}{dt}=kC_A
$$

The analytical solution is:

$$
C_A=C_{A0}e^{-kt}
$$

```python
k = 0.1
CA0 = 1.0

t = np.linspace(0,50,200)

CA = CA0*np.exp(-k*t)
```

```python
plt.figure(figsize=(8,4))

plt.plot(
    t,
    CA,
    label="First-order batch reaction"
)

plt.xlabel("Time")
plt.ylabel("CA")

plt.legend()
plt.grid(True)

plt.show()
```

## 4. Conversion profiles

For a batch reactor:

$$
X=\frac{C_{A0}-C_A}{C_{A0}}
$$

```python
X=(CA0-CA)/CA0
```

```python
plt.figure(figsize=(8,4))

plt.plot(
    t,
    X,
    label="Conversion"
)

plt.xlabel("Time")
plt.ylabel("Conversion")

plt.grid(True)
plt.legend()

plt.show()
```

## 5. Reaction rate profiles

The reaction rate can be calculated from:

$$
-r_A=kC_A
$$

```python
rate = k*CA
```

```python
plt.figure(figsize=(8,4))

plt.plot(
    CA,
    rate
)

plt.xlabel("CA")
plt.ylabel("-rA")

plt.grid(True)

plt.show()
```

## 6. Comparing reaction orders

```python
k1=0.1
k2=0.1

first_order=k1*CA
second_order=k2*CA**2
```

```python
plt.figure(figsize=(8,4))

plt.plot(
    CA,
    first_order,
    label="First order"
)

plt.plot(
    CA,
    second_order,
    label="Second order"
)

plt.xlabel("CA")
plt.ylabel("Rate")

plt.legend()
plt.grid(True)

plt.show()
```

## 7. Arrhenius plot

The Arrhenius equation is:

$$
k=Ae^{-E_a/RT}
$$

Taking logarithm:

$$
ln(k)=ln(A)-\frac{E_a}{R}\frac{1}{T}
$$

Therefore, plotting:

$$
ln(k) \ vs \ 1/T
$$

gives a straight line.

```python
R=8.314

T=np.array([300,320,340,360])

k=np.array([
    0.01,
    0.018,
    0.032,
    0.055
])

plt.figure(figsize=(8,4))

plt.plot(
    1/T,
    np.log(k),
    marker="o"
)

plt.xlabel("1/T")
plt.ylabel("ln(k)")

plt.grid(True)

plt.show()
```

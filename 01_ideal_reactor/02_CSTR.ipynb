# Continuous Stirred Tank Reactor (CSTR)

A Continuous Stirred Tank Reactor (CSTR) is one of the most important continuous reactor configurations in chemical reaction engineering.

In a CSTR, reactants are continuously fed into the reactor while products are continuously removed.

The reactor contents are assumed to be perfectly mixed, meaning that the composition inside the reactor is identical to the outlet stream.

CSTRs are widely used in:

- Large-scale chemical production
- Polymerization processes
- Wastewater treatment
- Biochemical production systems
- Continuous manufacturing processes

Unlike a batch reactor, where the reactor performance changes with time, an ideal CSTR operates at steady state.

The main engineering question changes from:

"How long should the reaction proceed?"

to:

"How large should the reactor volume be to achieve the desired conversion?"

This notebook introduces CSTR modeling through the workflow:

**Physical concept → Mathematical model → Python simulation → Parameter analysis → Reactor design**

## 1. Physical concept of a CSTR

A Continuous Stirred Tank Reactor (CSTR) is a continuous-flow reactor in which reactants enter the reactor and products leave continuously during operation.

The key feature of an ideal CSTR is **perfect mixing**.

Because of perfect mixing, the composition and temperature inside the reactor are uniform throughout the entire reactor volume.

Therefore:

$$
C_{A,out}=C_A
$$

where:

- $C_{A,out}$ is the outlet concentration of species A
- $C_A$ is the concentration inside the reactor

This assumption allows the reactor outlet stream to be directly related to the reactor conditions.

---

### Ideal CSTR assumptions

The mathematical model of an ideal CSTR is based on the following assumptions:

#### 1. Continuous operation

Reactants continuously enter the reactor, and products continuously leave the reactor.

$
F_{in}\neq0
$

$
F_{out}\neq0
$

---

#### 2. Perfect mixing

The reactor contents are completely mixed:

$
C_{reactor}=C_{out}
$

This means there are no concentration gradients inside the reactor.

---

#### 3. Steady-state operation

After sufficient operating time, the reactor reaches a condition where all variables remain constant with time.

Therefore:

$$
\frac{dN_A}{dt}=0
$$

---

### CSTR Compared with Batch Reactor

| Feature | Batch Reactor | CSTR |
|---|---|---|
| Feed stream | No | Yes |
| Outlet stream | No | Yes |
| Operation mode | Transient | Steady state |
| Main variable | Time | Reactor volume |
| Design question | Required reaction time | Required reactor size |

The difference in operating mode leads to different reactor design equations, even when the same reaction kinetics are used.

## 2. Reaction system definition

To develop the mathematical model of a CSTR, we first define the reaction system and reaction kinetics.


In this notebook, we consider the same irreversible first-order reaction used in the batch reactor model:

$$
A \rightarrow B
$$

The reaction rate is:

$$
-r_A=kC_A
$$

where:

- $-r_A$ is the rate of disappearance of reactant A
- $k$ is the reaction rate constant
- $C_A$ is the concentration of reactant A

Using the same reaction system allows us to compare different reactor configurations based on identical reaction kinetics.

---

### Flow variables in a CSTR

Unlike a batch reactor, a CSTR operates continuously.

Therefore, the material balance is described using molar flow rates:

$$
F_A=vC_A
$$

where:

- $F_A$ is the molar flow rate of species A
- $v$ is the volumetric flow rate
- $C_A$ is the concentration of species A

The inlet stream is represented by:

$$
F_{A0}=v_0C_{A0}
$$

where:

- $F_{A0}$ is the inlet molar flow rate
- $v_0$ is the inlet volumetric flow rate
- $C_{A0}$ is the inlet concentration of A

---

### Conversion

Conversion is an important design variable in reactor engineering.

For reactant A:

$$
X_A=
\frac{F_{A0}-F_A}{F_{A0}}
$$

Rearranging:

$$
F_A=F_{A0}(1-X_A)
$$

This relationship connects the material flow information with reactor performance.

In the following section, we will derive the CSTR design equation using a steady-state mole balance.

## 3. Steady-state mole balance derivation

The mathematical model of a CSTR is developed from the general mole balance equation:

$$
\text{Input}
-
\text{Output}
+
\text{Generation}
=
\text{Accumulation}
$$

For species A:

$$
F_{A0}
-
F_A
+
r_AV
=
\frac{dN_A}{dt}
$$

where:

- $F_{A0}$ is the inlet molar flow rate of A
- $F_A$ is the outlet molar flow rate of A
- $r_A$ is the rate of generation of A
- $V$ is the reactor volume
- $N_A$ is the number of moles of species A inside the reactor

---

### Applying CSTR assumptions

For an ideal steady-state CSTR:

#### 1. Steady-state operation

The amount of material inside the reactor does not change with time:

$$
\frac{dN_A}{dt}=0
$$

Therefore, the accumulation term disappears.

The mole balance becomes:

$$
0=
F_{A0}
-
F_A
+
r_AV
$$

---

### Rearranging the balance equation

Moving the reaction term to the other side:

$$
F_{A0}-F_A=-r_AV
$$

The difference between inlet and outlet flow represents the amount of reactant consumed:

$$
F_{A0}-F_A=F_{A0}X_A
$$

Therefore:

$$
F_{A0}X_A=-r_AV
$$

Solving for reactor volume:

$$
\boxed{
V=
\frac{F_{A0}X_A}{-r_A}
}
$$

This equation is known as the **CSTR design equation**.

---

### Physical meaning of the CSTR design equation

The CSTR design equation shows that reactor volume depends on:

#### Reaction demand

$$
F_{A0}X_A
$$

which represents the amount of reactant that must be converted.

#### Reaction capability

$$
-r_A
$$

which represents how fast the reaction consumes reactant A.

Therefore:

$$
\boxed{
\text{Required reactor volume}
=
\frac{\text{Required reaction amount}}
{\text{Reaction rate}}
}
$$

This relationship connects reaction kinetics with reactor design.

## 4. CSTR design equation for first-order reaction

The general CSTR design equation is:

$$
V=
\frac{F_{A0}X_A}{-r_A}
$$

This equation can be applied to different reaction kinetics.

In this section, we apply the design equation to the first-order irreversible reaction:

$$
A\rightarrow B
$$

with the rate expression:

$$
-r_A=kC_A
$$

---

### Relating reaction rate to conversion

For a constant-volume liquid-phase reaction:

$$
C_A=C_{A0}(1-X_A)
$$

Substituting into the rate expression:

$$
-r_A=kC_{A0}(1-X_A)
$$

Therefore, the CSTR design equation becomes:

$$
V=
\frac{F_{A0}X_A}
{kC_{A0}(1-X_A)}
$$

---

### Introducing volumetric flow rate

The inlet molar flow rate can be expressed as:

$$
F_{A0}=v_0C_{A0}
$$

where:

- $v_0$ is the inlet volumetric flow rate

Substituting:

$$
V=
\frac{v_0C_{A0}X_A}
{kC_{A0}(1-X_A)}
$$

The concentration terms cancel:

$$
\boxed{
V=
\frac{v_0X_A}
{k(1-X_A)}
}
$$

---

### Engineering meaning

The CSTR design equation shows how reactor volume depends on operating conditions:

#### Increasing conversion

$$
X_A\uparrow
$$

requires a larger reactor volume.

Because the remaining reactant concentration becomes lower, reducing the reaction rate.

---

#### Increasing reaction rate constant

$$
k\uparrow
$$

reduces the required reactor volume.

A faster reaction requires less reactor space to achieve the same conversion.

---

#### Increasing flow rate

$$
v_0\uparrow
$$

increases the required reactor volume.

A larger amount of material must be processed per unit time.

The design equation connects:

$$
\boxed{
\text{Kinetics}
+
\text{Flow conditions}
+
\text{Conversion target}
\rightarrow
\text{Reactor size}
}
$$

## 5. Python implementation: CSTR reactor sizing

In the previous sections, we derived the CSTR design equation:

$$
V=
\frac{v_0X_A}
{k(1-X_A)}
$$

This equation can be directly implemented in Python to calculate the required reactor volume.

In this section, we will define the operating parameters and build a simple CSTR design calculation.

The calculation workflow is:

$$
\text{Operating conditions}
\rightarrow
\text{Reaction kinetics}
\rightarrow
\text{Conversion target}
\rightarrow
\text{Reactor volume}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define CSTR operating conditions

v0 = 1.0      # Volumetric flow rate (L/min)
k = 0.5       # Reaction rate constant (1/min)
X = 0.8       # Desired conversion
```

```python
# Calculate required reactor volume

V = v0 * X / (k * (1 - X))
print(
    f"Required CSTR volume: {V:.2f} L"
)
```

## 6. Conversion analysis and reactor volume

The CSTR design equation describes the relationship between reactor volume and conversion:

$$
V=
\frac{v_0X_A}
{k(1-X_A)}
$$

Instead of calculating only one operating point, we can analyze how the required reactor volume changes with conversion.

This type of analysis is important for reactor design because increasing conversion usually requires a larger reactor volume.

In this section, we will investigate:

- How conversion affects reactor size
- Why very high conversion requires significantly larger reactors
- The engineering trade-off between conversion and reactor cost

```python
# Generate conversion range

X_values = np.linspace(0.01, 0.99, 200)
V_values = v0 * X_values / (k * (1 - X_values))
```

```python
plt.figure(figsize=(7,4))

plt.plot(
    X_values,
    V_values
)

plt.xlabel("Conversion")
plt.ylabel("Reactor Volume (L)")
plt.title("CSTR Reactor Volume vs Conversion")
plt.grid()

plt.show()
```

### Engineering Interpretation

The conversion-volume relationship shows an important characteristic of CSTR design.

As conversion increases:

$$
X_A \rightarrow 1
$$

the remaining concentration of reactant decreases:

$$
C_A=C_{A0}(1-X_A)
$$

which reduces the reaction rate:

$$
-r_A=kC_A
$$

Because the reaction becomes slower, a much larger reactor volume is required to achieve additional conversion.

Therefore, reactor design involves a trade-off between:

- Higher conversion
- Larger reactor volume
- Higher capital cost
- Improved product recovery

In industrial processes, the optimal conversion is usually determined by balancing reactor cost and downstream separation requirements.

## 7. Building a reusable CSTR design function

In the previous section, the CSTR volume was calculated using a fixed set of operating conditions.

However, engineering calculations often require evaluating many different scenarios.

To improve reusability, we convert the CSTR design equation into a Python function.

The function will calculate the required reactor volume based on:

- Desired conversion
- Reaction rate constant
- Volumetric flow rate

The mathematical model is:

$$
V=
\frac{v_0X_A}
{k(1-X_A)}
$$

This approach allows the reactor model to be reused for different design conditions.

```python
def cstr_volume(X, k, v0):
    """
    Calculate the required volume of an ideal CSTR.

    Parameters:
    X : float
        Desired conversion
    k : float
        First-order reaction rate constant (1/min)
    v0 : float
        Volumetric flow rate (L/min)

    Returns:
    V : float
        Required reactor volume (L)
    """

    V = v0 * X / (k * (1 - X))

    return V
```

```python
# Test the CSTR design function

X_test = 0.9

k_test = 0.5

v0_test = 1.0


V_test = cstr_volume(
    X_test,
    k_test,
    v0_test
)


print(
    f"Required CSTR volume: {V_test:.2f} L"
)
```

### Engineering interpretation

Creating a reusable function transforms the reactor model from a single calculation into a computational design tool.

The function allows engineers to quickly evaluate different operating scenarios:

- Increasing conversion requirements
- Changing reaction kinetics
- Scaling production flow rates

For example:

Higher conversion:

$$
X_A\uparrow
$$

requires a larger reactor volume.

Faster kinetics:

$$
k\uparrow
$$

reduces the required reactor volume.

Higher production rate:

$$
v_0\uparrow
$$

requires a larger reactor.

This function represents a simplified reactor design calculator that can be extended to more complex reactor models.

## 8. Parameter analysis

The CSTR design equation:

$$
V=
\frac{v_0X_A}
{k(1-X_A)}
$$

shows that reactor volume is controlled by several important operating parameters.

In this section, we investigate the influence of:

- Reaction kinetics
- Conversion requirement
- Production flow rate

on the required reactor volume.

Understanding these relationships is essential for reactor design and process optimization.

```python
# Effect of reaction rate constant

k_values = [0.1, 0.5, 1.0]

X_fixed = 0.8
v0_fixed = 1.0

for k_value in k_values:

    V = cstr_volume(
        X_fixed,
        k_value,
        v0_fixed
    )

    print(
        f"k = {k_value} 1/min --> V = {V:.2f} L"
    )
```

```python
# Effect of conversion

X_values = np.linspace(0.1,0.95,100)

k_fixed = 0.5
v0_fixed = 1.0

V_values = cstr_volume(
    X_values,
    k_fixed,
    v0_fixed
)

plt.figure(figsize=(7,4))

plt.plot(
    X_values,
    V_values
)

plt.xlabel("Conversion")
plt.ylabel("Reactor Volume (L)")
plt.title(
    "Effect of Conversion on CSTR Volume"
)
plt.grid()

plt.show()
```

```python
# Effect of volumetric flow rate

v0_values = [0.5, 1.0, 2.0]

X_fixed = 0.8
k_fixed = 0.5

for flow_rate in v0_values:

    V = cstr_volume(
        X_fixed,
        k_fixed,
        flow_rate
    )

    print(
        f"v0 = {flow_rate} L/min --> V = {V:.2f} L"
    )
```

### Engineering Interpretation

The parameter analysis demonstrates how operating conditions influence CSTR design.

#### 1. Effect of reaction kinetics

Increasing the reaction rate constant:

$$
k\uparrow
$$

increases reaction speed and decreases the required reactor volume.

This explains why catalysts are valuable in industrial reactors.

#### 2. Effect of conversion

Increasing conversion:

$$
X_A\uparrow
$$

requires a larger reactor volume.

At very high conversion, the remaining reactant concentration becomes very low, causing a significant decrease in reaction rate.

#### 3. Effect of flow rate

Increasing volumetric flow rate:

$$
v_0\uparrow
$$

means more material must be processed per unit time, resulting in a larger reactor requirement.

These relationships demonstrate that reactor design requires balancing:

$$
\boxed{
\text{Kinetics}
+
\text{Production Rate}
+
\text{Conversion Requirement}
}
$$
